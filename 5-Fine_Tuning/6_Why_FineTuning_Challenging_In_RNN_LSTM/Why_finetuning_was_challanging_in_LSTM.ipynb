```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Embedding, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
max_len = 200
vocab_size = 10000
embedding_dim = 128
latent_dim = 256 ## Hidden layer dim
```

```python
(x_train, y_train), _ = tf.keras.datasets.imdb.load_data(num_words=vocab_size)
```

```python
len(x_train)
len(y_train)
```

```python
x_train = x_train[:3000]
y_train = y_train[:3000]
```

```python
len(x_train)
len(y_train)
## Below shoukd be 3000 only
```

```python
x_train = pad_sequences(x_train, maxlen=max_len, padding='post', truncating='post')
```

```python
input_layer = Input(shape=(max_len,))
embedding_layer = Embedding(vocab_size, embedding_dim)(input_layer)
lstm_layer, state_h, state_c = LSTM(latent_dim, return_state=True)(embedding_layer)
output_layer = Dense(1, activation='sigmoid')(state_h)
classification_model = Model(input_layer, output_layer)
```

```python
classification_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
classification_model.fit(x_train, y_train, batch_size=64, epochs=1, validation_split=0.1)
```

## Retraining for Summarization

What we have done here is that we have taken the pretrained LSTM layer that we just trained and considered it as Encoder and Decoder we designed from our side and then we finetuned the resulting for the Summarization task.

```python
updated_classification_model = load_model("lstm_imdb_model_updated.h5")
```

```python
encoder_inputs = Input(shape=(max_len,))
encoder_embedding = updated_classification_model.layers[1](encoder_inputs)
encoder_outputs, state_h, state_c = updated_classification_model.layers[2](encoder_embedding)
```

```python
output_vocab_size = 8000
target_max_len = 50

decoder_inputs = Input(shape=(None,))
decoder_embedding = Embedding(output_vocab_size, embedding_dim)(decoder_inputs)
decoder_lstm = LSTM(latent_dim, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(decoder_embedding, initial_state=[state_h, state_c])
decoder_dense = Dense(output_vocab_size, activation='softmax')
decoder_outputs = decoder_dense(decoder_outputs)
```

```python
seq2seq_model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
seq2seq_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
encoder_input_data = x_train[:1000]  # using IMDB input for now
decoder_input_data = np.random.randint(1, output_vocab_size, (1000, target_max_len))
decoder_target_data = np.random.randint(1, output_vocab_size, (1000, target_max_len, 1))

## Recall in Decoder side we pass the Ground Truth data as input and the target data is shifted by one position. This is called Teacher Forcing.
## This is represented in the below code snippet. The decoder_input_data is the Ground Truth data and the decoder_target_data is the shifted version of it. This is done to help the model learn better during training.
```

```python
seq2seq_model.fit(
    [encoder_input_data, decoder_input_data],
    decoder_target_data,
    batch_size=32,
    epochs=1,
    validation_split=0.1
)
```

# Please note:
- Above first of all we used LSTM and kind of PreTrained it for Text Classification.
- Then we used the same LSTM for FineTuning over Text Summarization task.
- Now, using the same Pretrained model we can just do the Summarization task. Say for example if we intend to do the Classification task(This is what we initially we made our LSTM model to work upon) it won't be able to do that.
- Whereas, on the other transformer based architecture when finetuned for specific task still holds the capbility to perform the other NLP task which it has learned as part of its pretraining.
- This is one of the main reason why Fine Tuning is not possible in LSTM/RNN and how Trasformer changed the game.
- Also, the Fine Tuning frameworks like Hugging Face supports(coding wise) Fine Tuning of Transformers based LLMs like BERT but not LSTM/RNN.
- In LSTM based architecture scaling was not possible since it was timestamp based approach that is one word at a time whereas in transformers we can pass all the words parallely in the same timestamp therefore making them highly scallable compared to LSTM.
